# Análise

### Importar bibliotecas

```python
import pandas as pd
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import seaborn as sns
```

### Separação entre conjunto de testes e conjunto de treinamento

```python
df_tratado = pd.read_csv("../dataset/student.csv")

# Divisão Treino e Teste
# 30% para Teste e 70% para Treino
df_treino, df_teste = train_test_split(df_tratado, test_size=0.30, random_state=42)

# Separar a subamostra para exploração
# 80% aleatoriamente do conjunto de Treino.
df_eda = df_treino.sample(frac=0.80, random_state=42)

# Conferindo os tamanhos
print(f"Total na base limpa: {len(df_tratado)} alunos")
print(f"Conjunto de Teste: {len(df_teste)} alunos")
print(f"Conjunto de Treino: {len(df_treino)} alunos")
print(f"Subamostra para EDA: {len(df_eda)} alunos")
```
